## Generar una selección o el conjunto completo

La selección predeterminada comprueba familias distintas de gráficos. Cambiar `SELECCION=None` genera todos. Las entradas externas ausentes quedan identificadas en cobertura.json; los mapas permanecen excluidos.

```python
from datetime import datetime,timezone
SELECCION=[11,15,20,41,48]
PRIVADAS=None  # None usa entradas_privadas/ local, excluida de Git.
salida=BASE/"resultados"/("notebook_graficos_"+datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S_%fZ"))
generador=Graficos(salida,privadas=PRIVADAS)
cobertura=generador.generar(SELECCION)
pd.DataFrame(cobertura)
```
